```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
import numpy as np

# Define your feature matrix X and target y here
# X = ...
# y = ...

# Setting up the parameter grid to search over
param_grid = {
    'C': [0.1, 1, 10, 100],  # Regularization parameter
    'epsilon': [0.01, 0.1, 0.5, 1, 2],  # Epsilon in the epsilon-SVR model
      # Type of kernel function 
}

# Initialize the SVR model
svr = SVR()

# Set up GridSearchCV
grid_search = GridSearchCV(svr, param_grid, cv=5, scoring='neg_mean_squared_error', n_jobs=-1, verbose=2)

# Perform the grid search on your data
grid_search.fit(X, y)

# Best parameter set found:
print("Best parameters found:", grid_search.best_params_)

# Best score
print("Best score (neg_mean_squared_error):", grid_search.best_score_)

# Optionally, you can access the best estimator directly
best_svr_model = grid_search.best_estimator_

# Now, you can use best_svr_model for predictions or further analysis


```

```python
# Split the dataset
# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)



# Define pipeline components
standardizer = StandardScaler()
regression_model = SVR(C=100,  # Regularization parameter
                       epsilon=2,  # Epsilon in the epsilon-SVR model
                      )  # Kernel type# Epsilon in the epsilon-SVR model



 #=RandomForestRegressor(max_depth=None,  # or other optimal value
      #  min_samples_leaf=2,
       # min_samples_split=5,
       # n_estimators=100,
        #random_state=42) # Or Ridge(alpha=0.184)

pipeline = Pipeline([
    ('standardizer', StandardScaler()),
    ('pca', PCA(n_components = 200)),  # You can specify n_components here if desired
    ('regression_model',regression_model)]) 
    
# Setup cross-validation
cv = RepeatedKFold(n_splits=5, n_repeats=5, random_state=42)

# Perform cross-validation
cv_scores = cross_val_score(pipeline, X_train, y_train, cv=cv, scoring='neg_mean_squared_error')
cv_scores = -cv_scores  # Convert to positive values

# Print CV results
print(f"CV Mean Squared Error: {cv_scores.mean()} (± {cv_scores.std() * 2})")

# Fit the pipeline and make predictions
pipeline.fit(X_train, y_train)
y_train_pred = pipeline.predict(X_train)
y_test_pred = pipeline.predict(X_test)




import matplotlib.pyplot as plt
import seaborn as sns

# Set the size of the plots
plt.figure(figsize=(14, 6))

# Plotting the distribution of ages in the training set
plt.subplot(1, 2, 1)  # 1 row, 2 columns, first subplot
sns.histplot(y_train, kde=True, bins=30, color='blue', label='Training Set')
plt.title('Age Distribution in the Training Set')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.legend()

# Plotting the distribution of ages in the test set
plt.subplot(1, 2, 2)  # 1 row, 2 columns, second subplot
sns.histplot(y_test, kde=True, bins=30, color='green', label='Test Set')
plt.title('Age Distribution in the Test Set')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Calculate performance metrics
r2_train = r2_score(y_train, y_train_pred)
rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
mae_train = mean_absolute_error(y_train, y_train_pred)
r2_test = r2_score(y_test, y_test_pred)
rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))
mae_test = mean_absolute_error(y_test, y_test_pred)
```

```python
# Print performance metrics
print(f"Training set - R^2: {r2_train}, RMSE: {rmse_train}, MAE: {mae_train}")
print(f"Test set - R^2: {r2_test}, RMSE: {rmse_test}, MAE: {mae_test}")


# Plot Actual vs. Predicted values
fig, ax = plt.subplots(1, 2, figsize=(12, 6), sharey=True)
ax[0].scatter(y_train, y_train_pred, color='orange', alpha=0.5)
ax[0].plot([y_train.min(), y_train.max()], [y_train.min(), y_train.max()], 'k--', lw=2)
ax[0].set_title('Training Set: Actual vs. Predicted')
ax[0].set_xlabel('Actual Age')
ax[0].set_ylabel('Predicted Age')

ax[1].scatter(y_test, y_test_pred, color='red', alpha=0.5)
ax[1].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
ax[1].set_title('Test Set: Actual vs. Predicted')
ax[1].set_xlabel('Actual Age')
plt.tight_layout()
plt.show()




```
